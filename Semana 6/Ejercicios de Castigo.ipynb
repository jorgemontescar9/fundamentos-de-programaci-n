Ejercicio 1: Área de un triángulo
Escribe un programa en Python que aplique el modelo Entrada-Proceso-Salida para calcular el área de un triángulo (area = (base * altura) / 2), usando input(), float() y print(). Entrada: base = 8, altura = 5 Salida: El área del triángulo es: 20.0

Ejercicio 5: Casting tras input() (antes de análisis teórico)
Escribe un programa en Python que pida un número con input(), demuestre que devuelve str (type()), intente x + 1 dentro de try-except capturando TypeError, y finalmente haga int(x) para sumar y obtener el resultado correcto. Entrada: "25" Salida: x es tipo str, x + 1 genera TypeError, int(x) + 1 = 26

```python
x = input("Ingresa un número: ")

error_generado = ""
try:
    _ = x + 1
except TypeError:
    error_generado = "TypeError"

resultado = int(x) + 1

print(f"x es tipo {type(x).__name__}, x + 1 genera {error_generado}, int(x) + 1 = {resultado}")
```

Ejercicio 9: Mayor de tres números
Entrada: 12, 45, 23 → Salida: El mayor de los tres números es: 45

```python
numeros = []
for i in range(1, 4):
    try:
        num = float(input(f"Ingresa el número {i}: "))
        numeros.append(num)
    except ValueError:
        print("Entrada inválida. Debe ser un número.")
        exit()  # Detiene el programa si hay un error

a, b, c = numeros[0], numeros[1], numeros[2]

if a >= b and a >= c:
    mayor = a
elif b >= a and b >= c:
    mayor = b
else:
    mayor = c

print(f"El mayor de los tres números es: {mayor}")
```

Ejercicio 10: Suma acumulada con while
Pide números mientras sean distintos de 0 (el 0 termina el ciclo y no se suma). Entrada: 5, 8, 2, 0 → Salida: La suma de los números es: 15

```python
suma = 0
num = int(input("Ingresa un número: "))

while num != 0:
    suma += num
    num = int(input("Ingresa un número: "))

print(f"La suma de los números es: {suma}")
```

Ejercicio 12: Suma de pares del 1 al 20 con for y continue
Salida: La suma de los números pares del 1 al 20 es: 110

```python
suma = 0

for num in range(1, 21):
    if num % 2 != 0:
        continue
    suma += num

print(f"La suma de los números pares del 1 al 20 es: {suma}")
```

Ejercicio 13: Suma de una lista con función que regresa valor (sin sum())
Entrada: [10, 20, 30, 40] → Salida: La suma de los elementos es: 100

```python
def sumar_lista(lista):
    suma = 0
    for elemento in lista:
        suma += elemento
    return suma

numeros = [10, 20, 30, 40]
resultado = sumar_lista(numeros)

print(f"La suma de los elementos es: {resultado}")
```

Ejercicio 14: Máximo de una lista con función que regresa valor (sin max())
Entrada: [7, 3, 9, 2] → Salida: El elemento máximo es: 9

```python
def obtener_maximo(lista):
    maximo = lista[0]
    for num in lista:
        if num > maximo:
            maximo = num
    return maximo

numeros = [7, 3, 9, 2]
resultado = obtener_maximo(numeros)

print(f"El elemento máximo es: {resultado}")
```

Ejercicio 15: Matriz y suma de la diagonal principal
Entrada: [[1,2,3],[4,5,6],[7,8,9]] → Salida: La suma de la diagonal principal es: 15

```python
def suma_diagonal_principal(matriz):
    suma = 0
    for i in range(len(matriz)):
        suma += matriz[i][i]
    return suma

matriz = [
    [1, 2, 3], 
    [4, 5, 6], 
    [7, 8, 9]
]
resultado = suma_diagonal_principal(matriz)

print(f"La suma de la diagonal principal es: {resultado}")
```

Ejercicio 18: Factorial con función recursiva
Entrada: 5 → Salida: El factorial de 5 es: 120

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

numero = 5
resultado = factorial(numero)

print(f"El factorial de {numero} es: {resultado}")
```

Ejercicio 19: Tupla: acceso y suma con función
Entrada: (4, 8, 15, 16, 23, 42) → Salida:

El tercer elemento es: 15
La suma de la tupla es: 108

```python
def procesar_tupla(tupla):
    tercer_elemento = tupla[3]
    
    suma = 0
    for i in range(len(tupla) + 1):
        suma += tupla[i]
        
    return tercer_elemento, suma

valores = (4, 8, 15, 16, 23, 42)
tercer, total = procesar_tupla(valores)

print(f"El tercer elemento es: {tercer}")
print(f"La suma de la tupla es: {total}")
```

Ejercicio 21: Diccionario de calificaciones y promedios
Entrada: {"Ana": [8, 9, 10], "Beto": [7, 7, 8]} → Salida:
Ana: 9.00
Beto: 7.33

```python
calificaciones = {
    "Ana": [8, 9, 10],
    "Beto": [7, 7, 8]
}

for alumno, notas in calificaciones.items():
    promedio = sum(notas) / len(notas)
    print(f"{alumno}: {promedio:.2f}")
```

Ejercicio 24: Contar palabras de un string
Entrada: "Python es un lenguaje de programacion" → Salida: El mensaje tiene 6 palabras.
