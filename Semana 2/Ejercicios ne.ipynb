Ejercicio 1: Mayor de edad
Enunciado: Crea un programa que pida la edad de una persona y muestre si es mayor de edad (18 años o más) o menor de edad. Entrada: 17 Salida: Eres menor de edad

```python
edad = int(input("Ingresa tu edad"))
if edad <= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
    
```

**Ejercicio 2:** Número par o impar


Enunciado: Crea un programa que pida un número entero y determine si es par o impar usando el operador módulo %. Entrada: 7 Salida: 7 es un número impar

```python
numero = int(input("Ingresa un número entero"))

if (numero % 2 == 0):
    print("Es un número par")
else:
    print("Es un número impar")
```

Ejercicio 3: Calificación aprobatoria
Enunciado: Crea un programa que pida una calificación de 0 a 10 y muestre si la materia está aprobada (mayor o igual a 6) o reprobada. Entrada: 7 Salida: Aprobado

```python
numero = int(input("Ingresa un número de 0 a 10"))
if (numero >= 6):
    print("Aprovado")
else:
    print("Reprobado")
```
