```python
distancia = int(input("Ingresa la distancia en kilometros: "))
tiempo = int(input("Ingresa el tiempo en horas: "))

input("Ingresa la distancia en kilometros")
input("Ingresa el tiempo en horas")

print("El dato de distancia es de tipo: ", type(distancia))
print("El dato de tiempo es de tipo: ", type(tiempo))

velocidad = distancia / tiempo

print("La velocidad promedio es: ", velocidad)
```
