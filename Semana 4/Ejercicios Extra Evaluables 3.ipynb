Extra 1: Suma de todos los elementos de una matriz
Enunciado: Crea una función suma_matriz(matriz) que regrese valor (con return) la suma de todos los elementos de una lista de listas, usando ciclos for anidados. Define además una función imprimir_matriz(matriz) que no regrese valor y muestre la matriz sin corchetes ni comas. Aplica ambas funciones a la matriz [[1, 2, 3], [4, 5, 6], [7, 8, 9]]. Entrada:

Matriz: [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

```python
def suma_matriz(matriz):
    total = 0
    for columna in matriz:
        for elemento in columna:
            total += elemento
    return total

def imprimir_matriz(matriz):
    for columna in matriz:
        for elemento in columna:
            print(elemento, end=" ")
        print()

matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

imprimir_matriz(matriz)
resultado = suma_matriz(matriz)
print("La suma de todos los elementos de la matriz:", resultado)
```

Extra 2: Matriz con suma por fila
Enunciado: Crea una función suma_fila(fila) que regrese valor (con return) la suma de los elementos de una lista, y una función mostrar_tabla(matriz) que no regrese valor e imprima la matriz sin corchetes ni comas. Dada la matriz [[3, 1, 4], [1, 5, 9], [2, 6, 5]], imprime la matriz y la suma de cada fila. Entrada:

Matriz: [[3, 1, 4], [1, 5, 9], [2, 6, 5]]

```python
matriz = [
    [3,1,4],
    [1,5,9],
    [2,6,5]
]
for f in range(len(matriz)):
    for c in range(len(matriz[f])):
        print(matriz[f][c], end=" ")
    print()

for f in range(len(matriz)):
    suma = 0
    for c in range(len(matriz[f])):
        suma = suma + matriz[f][c]
    print("Suma de la fila", f"{f}:", suma)
```

Extra 3: Producto de una columna sin usar el operador *
Enunciado: Crea la matriz [[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]] (4 filas y 3 columnas). Define una función multiplicar_columna(matriz, columna) que regrese valor (con return) el producto de los elementos de una columna dada. Restricción: dentro de la función está prohibido usar el operador *; el producto debe calcularse con sumas repetidas usando un ciclo. Pide al usuario el número de columna (0, 1 o 2) y muestra el resultado. Entrada:

Numero de columna: 1

```python
#Matriz
matriz= [
    [1,2,3],
    [4,5,6],
    [7,8,9],
    [2,4,6]
]
def multiplicar_columna(matriz,columna):
    elementos = [fila[columna] for fila in matriz]
    resultado = elementos [0]
    for numero in elementos[1:]:
        acumulado = 0
        for _ in range(numero):
            acumulado += resultado
        resultado = acumulado
    return resultado

col = int(input("Ingresa el número de columna: "))
print(f"El producto de la columna {col} es: {multiplicar_columna(matriz, col)}")
```

Extra 4: Tabla de Pitágoras de tamaño variable
Enunciado: Crea una función generar_tabla(tamano) que regrese valor (con return) una lista de listas de tamano x tamano con los productos de la tabla de Pitágoras (renglones y columnas del 1 al tamano). Dentro de la función de generación está prohibido usar el operador *; usa sumas repetidas con ciclos. Define una función imprimir_tabla(tabla) que no regrese valor e imprima la tabla sin corchetes ni comas, con los números tabulados (\t). Pide al usuario un tamaño entre 2 y 5; si ingresa un valor fuera del rango, muestra un mensaje de error. Al final pide un renglón y una columna y muestra el producto consultando la matriz, sin usar *. Entrada:

Tamano de la tabla: 4
Renglon (factor): 2
Columna (factor): 3

```python
def generar_tabla(tamaño):
    tabla = []
    for i in range(1, tamaño + 1):
        renglon = []
        for j in range(1, tamaño + 1):
            renglon.append(i * j)
        tabla.append(renglon)
    return tabla

def imprimir_tabla(tabla):
    for fila in tabla:
        for elemento in fila:
            print(elemento, end="\t")
        print()

tamaño = int(input("Tamano de la tabla: "))
if tamaño < 2 or tamaño > 5:
    print("Error: el tamaño debe estar entre 2 y 5")
else:
    tabla = generar_tabla(tamaño)
    imprimir_tabla(tabla)
    renglon = int(input("Renglon (factor 1): "))
    columna = int(input("Columna (factor 2): "))
    
    producto = tabla[renglon - 1][columna - 1]
    print("El producto de", renglon, "x", columna, "es:", producto)
```
