Desafío Extra 1: Lista de espera con tuplas
Crea una tupla espera = ("María", "José", "Carlos", "Lucía", "Pedro"). Imprime el tercer elemento, captura dos nombres nuevos con input(), anéxalos en una nueva tupla, conviértela en lista, ordénala alfabéticamente y muestra la cantidad total de personas.

```python
espera = ("María", "José", "Carlos", "Lucía", "Pedro")

print("Tercer elemento:", espera[2])

nombre1 = input("Ingresa el primer nuevo nombre: ")
nombre2 = input("Ingresa el segundo nuevo nombre: ")
nueva_espera = espera + (nombre1, nombre2)

lista_espera = list(nueva_espera)
lista_espera.sort()

print("Lista ordenada alfabéticamente:", lista_espera)
print("Cantidad total de personas:", len(lista_espera))
```

Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario votos = {"Rojo": 0, "Azul": 0, "Verde": 0}. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con items() el color ganador junto con su total.

```python
votos = {"Rojo": 0, "Azul": 0, "Verde": 0}
for i in range(5):
    voto = input("Ingresa tu voto (Rojo, Azul, Verde): ").strip().capitalize()
    if voto in votos:
        votos[voto] += 1
    else:
        print("Esta opcion no es valida, no se tomara en cuenta este voto.")

ganador = ""
votos_totales = -1

for color, total in votos.items():
    print(f"{color}: {total} votos")
    if total > votos_totales:
        votos_totales = total
        ganador = color

print(f"El color ganador es {ganador} con {votos_totales} votos.")
```

Desafío Extra 3: Validación robusta de entrada
Escribe un programa que pida al usuario dos números enteros dentro de un bloque try-except. Si el usuario ingresa texto o un valor vacío, muestra un mensaje controlado y repite la solicitud con un ciclo while. Luego pide la división del primero entre el segundo, con un bloque específico para ZeroDivisionError.

```python
num1 = None
num2 = None

while True:
    try:
        num1 = int(input("Ingresa el primer número entero: "))
        num2 = int(input("Ingresa el segundo número entero: "))
        break
    except ValueError:
        print("Error, solo ingresa numeros enteros.")

try:
    resultado = num1 / num2
    print(f"{num1} dividido entre {num2} es: {resultado}")
except ZeroDivisionError:
    print("Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.")
```

Desafío Extra 4: Analizador de textos
Crea un programa que reciba una frase, imprima su longitud con len(), la convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre cuántas palabras contiene usando una función que retorne el conteo.

```python
def contar_palabras(texto):
    return len(texto.split())
frase = input("Ingresa una frase: ")

print("Longitud del mensaje:", len(frase))
print("En mayúsculas:", frase.upper())

frase_modificada = frase.replace("Python", "programación")
print("Texto reemplazado:", frase_modificada)

total_palabras = contar_palabras(frase)
print("Palabras totales:", total_palabras)
```

Desafío Extra 5: Minimenú modular integrador
Construye un menú con un ciclo while y opciones numéricas que permita ejecutar de forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en el diccionario contactos, (3) dividir dos números con manejo de ZeroDivisionError y (4) analizar un mensaje de texto. La opción (5) debe finalizar el programa.

```python
def ordenar_tupla(tupla_numeros):
    lista_temp = list(tupla_numeros)
    lista_temp.sort()
    return lista_temp

def buscar_telefono(diccionario, nombre):
    return diccionario.get(nombre, "Contacto no encontrado")

def dividir_numeros(num1, num2):
    try:
        resultado = num1 / num2
        return f"{num1} dividido entre {num2} es: {resultado}"
    except ZeroDivisionError:
        return "Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0."

def contar_palabras(texto):
    return len(texto.split())

def seccion_tuplas():
    print("\n--- SECCION TUPLAS ---")
    numeros = (10, 25, 30, 45, 50)
    print("Tupla original:", numeros)
    lista_ordenada = ordenar_tupla(numeros)
    print("Lista ordenada:", lista_ordenada)

def seccion_diccionarios():
    print("\n--- SECCION DICCIONARIOS ---")
    contactos = {
        "Ana": "555-0101",
        "Luis": "555-0102",
        "Mia": "555-0103"
    }
    nombre_buscar = input("Nombre a buscar: ")
    telefono = buscar_telefono(contactos, nombre_buscar)
    print("El telefono es:", telefono)

def seccion_excepciones():
    print("\n--- SECCION EXCEPCIONES ---")
    try:
        num1 = int(input("Primer numero entero: "))
        num2 = int(input("Segundo numero entero: "))
        mensaje = dividir_numeros(num1, num2)
        print(mensaje)
    except ValueError:
        print("Error: Tienes que meter numeros enteros.")

def seccion_strings():
    print("\n--- SECCION STRINGS ---")
    mensaje = input("Ingresa tu mensaje de texto: ")
    print("Longitud del mensaje:", len(mensaje))
    print("En mayusculas:", mensaje.upper())
    
    mensaje_modificado = mensaje.replace("Python", "programacion")
    print("Texto reemplazado:", mensaje_modificado)
    
    total_palabras = contar_palabras(mensaje)
    print("Palabras totales:", total_palabras)


def menu_principal():
    opcion = ""
    while opcion != "5":
        print("\n--- MENÚ PRINCIPAL ---")
        print("1. Mostrar tupla ordenada")
        print("2. Buscar teléfono en contactos")
        print("3. Dividir dos números")
        print("4. Analizar mensaje de texto")
        print("5. Finalizar")
        
        opcion = input("Selecciona una opción (1-5): ")
        
        if opcion == "1":
            seccion_tuplas()
        elif opcion == "2":
            seccion_diccionarios()
        elif opcion == "3":
            seccion_excepciones()
        elif opcion == "4":
            seccion_strings()
        elif opcion == "5":
            print("\nPrograma terminado.")
        else:
            print("Opcion no valida, pon un numero del 1 al 5.")

menu_principal()
```
