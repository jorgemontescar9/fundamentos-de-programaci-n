```python
#Codigo pa votar
candidatos = {
    "A": 0,
    "B": 0,
    "C": 0
}

cantidad = int(input("¿Cuantos votos van a ser?"))
conteo = 1

while conteo <= cantidad:
    voto = input(f"Ingresa el voto {conteo} (A, B o C): ")
    if voto.upper() in candidatos:
        print(f"Voto valido para {voto.upper()}")
        candidatos[voto.upper()] += 1
        conteo += 1
    else:
        print("El voto no es valido, vota nuevamente")

print("\nResultados:")
for k, v in candidatos.items():
    print(f"opcion {k}, votos: {v}")



    
     
```
