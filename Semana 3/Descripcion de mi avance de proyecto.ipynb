Sistema de Gestión de Inventario y Descuentos para Tienda de Artículos Deportivos 
Jorge Leonardo Montes Cardona
Avance de proyecto - Fundamentos de Programación
Cometas Athletics


Descripción del proyecto: Este proyecto nos va a ayudar en el sector de tiendas en general pero más específico para una tienda de artículos deportivos, a tener más organización en sus inventarios y un sistema que ayude con los descuentos de sus artículos determinando los directamente sin tener que revisar cuántos hay en inventario.

Problema a resolver: normalmente los cálculos hechos a mano al aplicar descuentos en gran volumen a inventarios puede generar errores en los valores finales de los artículos y también se pueden generar fallas al contabilizar las unidades restantes de cada artículo.

Como se va a resolver: voy a desarrollar un programa que automáticamente capture los datos de los productos, que éste valide si este articulo esta en stock y aplique descuentos según el volumen que se solicite, y vaya actualizando el inventario cada vez que este sufra un cambio.

Como operan actualmente estos negocios:
Actualización de ventas por medio de papel: Normalmente el encargado del establecimiento se encarga de contabilizar manualmente las ventas y el stock disponible.
Descuentos: Normalmente, estos se hacen conforme el vendedor lo considere, ya que en estos casos ellos deciden qué descuentos aplicar y cómo hacerlo, lo cual hace que no tengan algo establecido.
Inventario: Los trabajadores tienen que ir físicamente al inventario para poder corroborar que el producto esté en stock y poder verificar la cantidad que hay.


Reglas del negocio:
Restricción por cantidad: Este sistema no permitirá vender más cantidad de producto del que se encuentra en stock, si este se llega a sobrepasar el sistema mostrará un error en automático.
Regla de descuento por volumen: Si la cantidad de productos vendido es mayor a 5, en automático el sistema va a aplicar un descuento del 10% al total de la venta.
Regla de descuento por stock: Si la venta representa el 50% o más del stock total del producto el sistema automáticamente va a aplicar un descuento del 5% extra a su total.
Regla de actualización de stock: Tras confirmar que la venta se completó de manera correcta, el sistema debe verificar la cantidad de unidades vendidas y restantes en el stock.

Requerimientos Funcionales:
Nombre del producto (str): descripción del artículo en el inventario.
Costo por unidad (float): Costo de cada unidad.
Stock inicial (int): Cantidad de piezas inicial en el stock.
Cantidad a retirar (int): Unidades que serán vendidas.
Porcentaje de descuento (float): Porcentaje acumulado de descuento.
Monto descontado (float): Dinero restado al total de la venta.
Subtotal (float): Importe antes del descuento.
Total (float): Total a pagar.
Stock final (int): Stock restante después de la venta.
Validación de existencia: Comprobar que el producto a vender no supere el stock de este.
Operadores del Lenguaje
Matemáticos:
Multiplicación (*): Para calcular el importe base.
Resta (-) y Suma (+): Para acumuladores y desgloses finales.
División (/): Para obtener proporciones y porcentajes de descuento.
Relacionales:
Mayor o igual que (>=): Para evaluar si la cantidad es igual o supera las 5 unidades.
Igualdad (==): Para validación de menús y opciones del usuario.
Lógicos:
and: Para validar stock y condiciones de membresía.
not / or: Para control de flujos de salida en los ciclos.

Modelo Entrada - Proceso - Salida (EPS)
Entrada:
Nombre del producto, precio del producto y cantidad deseada.
Proceso:
Ingresar producto, valor y cantidad
Evaluar si cantidad >= 5 para aplicar 10% de descuento.
Acumular importes en el subtotal.
Aplicar descuento de (5%) si el total de artículos > al 50% del stock.
Salida:
Resumen de la compra.
Stock disponible
