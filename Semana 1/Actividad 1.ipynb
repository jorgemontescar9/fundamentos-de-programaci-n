Primero pregunte el total de la cuenta, y le asigne un nombre.
Despues pregunte el porcentaje de propina que el grupo queria dejar, y le asigne un nombre.
Despues pregunte el numero de personas y les asigne un nombre.
Despues hice los calculos de la cuenta por el porcentaje de propina, para tener la cantidad final y esta solo la dividi entre la cantidad de personas.
Y al final solo agregue los print para el mensaje de cuanto era la propina conforme a la cuenta, el total ya con propina y cuanto les tocaba a cada uno.

```python
#Ejercicios extra evaluables
    #Extra 1: Division de cuenta con propopina.
Total_de_la_cuenta = float(input("Total de la cuenta: "))
Porcentaje_de_propina = float(input("Porcentaje de propina: "))
Numero_de_personas = float(input("Numero de personas: "))

Propina = Total_de_la_cuenta * (Porcentaje_de_propina / 100)
Total_con_propina = Total_de_la_cuenta + Propina
Pago_por_persona = Total_con_propina / Numero_de_personas

print(f"La propina es: ${Propina:.2f}")
print(f"El total con propina es: ${Total_con_propina:.2f}")
print(f"Cada persona debe pagar: ${Pago_por_persona:.2f}")
```

 Aqui primero pregunte el total de minutos que tienen, con int ya que son minutos completos.
 Despues hice las operacion de los dias, horas y minutos definiendo cuanto tenia de cada uno al final.

```python
    #Extra 2: Conversor de minutos a días, horas y miutos
Total_de_minutos = int(input("Ingrese el total de minutos: "))
Dias = Total_de_minutos // 1440
Horas = (Total_de_minutos % 1440) // 60
Minutos = Total_de_minutos % 60

print(f"El total es: {Dias} días(s), {Horas} horas(s) y {Minutos} minutos(s)")
```

Primero pedi la cantidad de dinero en pesos mexicanos.
Despues hice la convercion en dolares de la cantidad en pesos / 18.5 como valor de dolar.
Despues hice la convercion en euros de la cantidad en pesos / 21 como valor del euro.
Y por ultimo utilice los print para arrojar que la cantidad en pesos se convertia a las otras monedas, (para esto utilice 2f para que me diera los 2 decimales que necesitaba) 

```python
Cantidad_en_MXN = float(input("Ingrese la cantidad en MXN: "))
Cambio_a_USD = Cantidad_en_MXN / 18.5
Cambio_a_EUR = Cantidad_en_MXN / 21
print("$1000.00 MXN equivalen a:")
print(f"El cambio a USD es: {Cambio_a_USD:.2f}")
print(f"El cambio a EUR es: {Cambio_a_EUR:.2f}")
```
