```python
#Calculadora1
Nombre_de_usuario = input("Ingrese su usuario: ")
num_1 = float(input("¿Cuantas horas pasas en redes sociales?: "))
num_2 = float(input("¿Cuantas horas pasas en videojuegos?: "))
num_3 = float(input("¿Cuantas horas pasas mandando mensajes?: "))
num_4 = float(input("¿Cuantas horas pasas estudiando en tu telefono?: "))
num_5 = float(input("¿Cuantas horas pasas viendo videos?: "))
tiempo_total = num_1 + num_2 + num_3 + num_4 + num_5
print(f"Hola, {Nombre_de_usuario}, pasas un {tiempo_total} de horas en tu telefono, el cual se traduce en porcetanje {tiempo_total / 24 * 100}%.")
```

```python
#Ejercicios extra evaluables
    #Extra 1: Division de cuenta con propopina.
Total_de_la_cuenta = float(input("Total de la cuenta: "))
Porcentaje_de_propina = float(input("Porcentaje de propina: "))
Numero_de_personas = float(input("Numero de personas: "))

Propina = Total_de_la_cuenta * (Porcentaje_de_propina / 100)
Total_con_propina = Total_de_la_cuenta + Propina
Pago_por_persona = Total_con_propina / Numero_de_personas

print(f"La propina es: ${Propina:.2f}")
print(f"El total con propina es: ${Total_con_propina:.2f}")
print(f"Cada persona debe pagar: ${Pago_por_persona:.2f}")
```

```python
    #Extra 2: Conversor de minutos a días, horas y miutos
Total_de_minutos = int(input("Ingrese el total de minutos: "))
Dias = Total_de_minutos // 1440
Horas = (Total_de_minutos % 1440) // 60
Minutos = Total_de_minutos % 60

print(f"El total es: {Dias} días(s), {Horas} horas(s) y {Minutos} minutos(s)")
```
